```python
from ml_advisor.utils.ollama_control import OllamaSession

with OllamaSession(auto_close=False) as h:
    response = httpx.post(
        f"{h}/api/generate",
        json={
            "model": "mistral:7b-instruct-q4_K_M",
            "prompt": "Em uma frase, o que é um teste t de Student?",
            "stream": False,
        },
        timeout=300,
        trust_env=False,
    )
    response.raise_for_status()

print(response.json()["response"])
```
